# Customer Support Ticket System - Data Preprocessing Notebook

This notebook demonstrates the text cleaning, tokenization, stopword removal, lemmatization, and TF-IDF feature extraction pipeline.

```python
import os
import pandas as pd
import sys
sys.path.append('..')
from utils.preprocess import clean_text, preprocess_text
from sklearn.feature_extraction.text import TfidfVectorizer
```

## 1. Load Dataset

```python
df = pd.read_csv('../dataset/tickets.csv')
print("Dataset Shape:", df.shape)
df.head()
```

## 2. Text Preprocessing Pipeline Walkthrough

```python
sample_texts = [
    "My credit card was charged $99.99 twice for order #10023! Check https://example.com/receipt",
    "URGENT: My internet is DOWN and router lights are RED. Please help asap!",
    "I want a full refund for defective item PRO-992. Delivered 3 days ago."
]

for i, raw in enumerate(sample_texts, 1):
    cleaned = clean_text(raw)
    processed = preprocess_text(raw)
    print(f"Sample {i}:")
    print(" RAW       :", raw)
    print(" CLEANED   :", cleaned)
    print(" PROCESSED :", processed)
    print("-" * 70)
```

## 3. Preprocess Entire Corpus

```python
df['processed_text'] = df['ticket_text'].apply(preprocess_text)
df[['ticket_text', 'processed_text']].head(10)
```

## 4. Vectorization (TF-IDF)

```python
vectorizer = TfidfVectorizer(max_features=10, ngram_range=(1, 2))
tfidf_matrix = vectorizer.fit_transform(df['processed_text'])
feature_names = vectorizer.get_feature_names_out()

tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)
print("TF-IDF Sample Features:", feature_names)
tfidf_df.head()
```
