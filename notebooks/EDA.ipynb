# Customer Support Ticket System - Exploratory Data Analysis (EDA)

This notebook analyzes the category distribution, priority distribution, department mapping, ticket length, and top n-gram words.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import sys
sys.path.append('..')
from utils.preprocess import preprocess_text

plt.style.use('ggplot')
```

## 1. Dataset Summary & Missing Values

```python
df = pd.read_csv('../dataset/tickets.csv')
print("Dataset Shape:", df.shape)
print("\nDataset Info:")
df.info()
print("\nMissing Values:\n", df.isnull().sum())
print("\nDuplicate Tickets:", df.duplicated(subset=['ticket_text']).sum())
df.head()
```

## 2. Target Class Distributions

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Category
sns.countplot(data=df, y='category', order=df['category'].value_counts().index, ax=axes[0], palette='viridis')
axes[0].set_title('Category Distribution')

# Priority
sns.countplot(data=df, x='priority', order=['Low', 'Medium', 'High', 'Critical'], ax=axes[1], palette='magma')
axes[1].set_title('Priority Distribution')

# Department
sns.countplot(data=df, y='department', order=df['department'].value_counts().index, ax=axes[2], palette='plasma')
axes[2].set_title('Department Distribution')

plt.tight_layout()
plt.show()
```

## 3. Category vs Priority Crosstab

```python
crosstab = pd.crosstab(df['category'], df['priority'])
plt.figure(figsize=(10, 6))
sns.heatmap(crosstab, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Category vs Priority Breakdown')
plt.show()
```

## 4. Ticket Text Length Analysis

```python
df['char_length'] = df['ticket_text'].astype(str).apply(len)
df['word_count'] = df['ticket_text'].astype(str).apply(lambda x: len(x.split()))

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
sns.histplot(df['word_count'], bins=20, kde=True, color='purple')
plt.title('Word Count Distribution')

plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='word_count', y='category', palette='Set3')
plt.title('Word Count by Category')

plt.tight_layout()
plt.show()
```

## 5. Most Frequent Words Analysis

```python
df['processed'] = df['ticket_text'].apply(preprocess_text)
all_words = ' '.join(df['processed']).split()
most_common = Counter(all_words).most_common(20)

words, counts = zip(*most_common)
plt.figure(figsize=(12, 5))
sns.barplot(x=list(counts), y=list(words), palette='mako')
plt.title('Top 20 Most Frequent Words across Corpus')
plt.xlabel('Frequency')
plt.show()
```
