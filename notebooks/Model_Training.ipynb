# Customer Support Ticket System - Model Training & Evaluation Notebook

This notebook trains and compares four classification algorithms (Logistic Regression, Naive Bayes, Random Forest, Linear SVM) on TF-IDF features.

```python
import os
import sys
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append('..')
from scripts.train_models import train_and_evaluate_models
```

## 1. Execute Training and Benchmark Models

```python
metrics = train_and_evaluate_models()
print("Selected Model:", metrics['best_category_model_name'])
print("Best F1 Score:", metrics['best_category_f1'])
```

## 2. Compare Model Performance Metrics

```python
comparison = metrics['models_comparison']
df_comp = pd.DataFrame(comparison).T[['accuracy', 'precision', 'recall', 'f1_score']]
df_comp
```

## 3. Visualize Accuracy & F1 Comparisons

```python
df_comp.plot(kind='bar', figsize=(10, 6))
plt.title('Classifier Performance Metrics Comparison')
plt.ylabel('Score')
plt.xticks(rotation=0)
plt.ylim(0.8, 1.05)
plt.legend(loc='lower right')
plt.show()
```

## 4. Confusion Matrix Visualization

```python
best_name = metrics['best_category_model_name']
cm = np.array(comparison[best_name]['confusion_matrix'])

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title(f'Confusion Matrix - {best_name}')
plt.xlabel('Predicted Category')
plt.ylabel('True Category')
plt.show()
```
